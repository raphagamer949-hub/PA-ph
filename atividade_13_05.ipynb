```python
# 5 soma, o usuario digita um numero e o programa soma do zero ate o numero

numero = int(input("Digite um número para somar até ele: "))

soma = 0
for i in range(numero + 1):
    soma += i

print(f"A soma de 0 até {numero} é: {soma}")
```

```python
# 8 contar vogais, o usuario digita uma palavra e o programa conta as vogais

palavra = input("Digite uma palavra: ")
vogais = "aeiou"
vogau = 0

for letra in palavra:
    if letra in vogais:
        vogau += 1

print(f"A palavra {palavra} tem {vogau} vogais.")
```

```python
# 9 maior numero digitado, o usuario digita quatro numeros e o programa imprime o maior numero digitado

print("Por favor, digite quatro números.")

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))
num4 = float(input("Digite o quarto número: "))

numeros = [num1, num2, num3, num4]

maior_numero = numeros[0]

for numero in numeros:
    if numero > maior_numero:
        maior_numero = numero

print(f"O maior número digitado é: {maior_numero}")
```
