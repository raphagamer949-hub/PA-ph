```python
#Jogo menu inteligente

#Criar um sistema que funcione até o usuário escolher sair

#Menu

# 1 - par/impar
# 2 - positivo/negativo
# 3 - maior
# 4 - sair


opcao = 0

while opcao != 4:
  print("--- Bem Vindo! ---")
  print("1 - Par / Ímpar")
  print("2 - Positivo / Negativo")
  print("3 - Maior")
  print("4 - Sair")
  print("------------")

  opcao = int(input("Escolha uma opção: "))

  if opcao == 1:
      numero = int(input("Digite um número: "))
      if numero % 2 == 0:
          print(f"O número {numero} é par")
      else:
          print(f"O número {numero} é ímpar")
  elif opcao == 2:
      numero = int(input("Digite um número para verificar (positivo/negativo): "))
      if numero > 0:
          print(f"O número {numero} é positivo.")
      elif numero < 0:
          print(f"O número {numero} é negativo.")
      else:
          print(f"O número {numero} é zero.")
  elif opcao == 3:
      num1 = int(input("Digite o primeiro número: "))
      num2 = int(input("Digite o segundo número: "))
      if num1 > num2:
          print(f"O maior número é: {num1}")
      elif num2 > num1:
          print(f"O maior número é: {num2}")
      else:
          print("Ambos os números são iguais.")
  elif opcao == 4:
      print("Saindo do programa. Até mais!")
  else:
      print("Opção inválida. Por favor, escolha um número de 1 a 4.")
```
