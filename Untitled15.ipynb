```python
#9

matriz = [
    [12, 5, 8],
    [20, 3, 15],
    [7, 25, 10]
]

maior = matriz[0][0]

for linha in matriz:
    for numero in linha:
        if numero > maior:
            maior = numero

print(maior)
```

```python
#10

notas = [
    [7, 8, 9],
    [6, 5, 8],
    [10, 9, 7]
]

# a) Notas do primeiro aluno
print("a)", notas[0])

# b) Segunda nota do segundo aluno
print("b)", notas[1][1])

# c) Terceira nota de cada aluno
print("c)")

for aluno in notas:
    print(aluno[2])

# d) Alterando a primeira nota do terceiro aluno
notas[2][0] = 8

print("d)", notas)
```
