```python
# QUESTÃO 02

numero = int(input("Digite um número para somar até ele: "))

soma = 0
for i in range(numero + 1):
    soma += i

print(f"A soma de 0 até {numero} é: {soma}")
```

```python
# QUESTÃO 03

numero = int(input("Digite um número para ver a tabuada: "))

print(f"Tabuada do {numero}:")
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

```python
# QUESTÃO 04

numero = -1
while numero < 0:
    try:
        numero = int(input("Digite um número inteiro positivo para calcular o fatorial: "))
        if numero < 0:
            print("O número deve ser positivo. Tente novamente.")
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")


fatorial = 1
for i in range(1, numero + 1):
    fatorial *= i
print(f"O fatorial de {numero} é: {fatorial}")
```

```python
# QUESTÃO 05

valor_saque = int(input("Digite o valor do saque: "))

notas = (100, 50, 20, 10, 5, 2)

print("Saída esperada:")

valor_restante = valor_saque

for nota in notas:
    if valor_restante >= nota:
        quantidade_notas = valor_restante // nota
        if quantidade_notas > 0:
            print(f"{quantidade_notas} nota(s) de R${nota}")
        valor_restante %= nota

if valor_restante > 0:
    print(f"Não foi possível sacar o valor exato. Sobrou R${valor_restante}.")
```
